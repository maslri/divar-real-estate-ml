```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
df = pd.read_csv("divar.csv",low_memory = False)
```

```python
mask = df[df["cat2_slug"].isin(['residential-sell','commercial-sell'])].copy()
```

```python
median_price = (mask.groupby("cat3_slug")["price_value"].median().sort_values(ascending=False))
plt.figure(figsize=(12,6))
median_price.plot(kind="bar")
plt.xticks(rotation=45)
plt.xlabel("Category")
plt.ylabel("Median Price")
plt.title("Median Price of Property Categories")
plt.show()
```
