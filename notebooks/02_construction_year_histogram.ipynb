```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv("divar.csv",low_memory = False)
```

```python
df["construction_year"] = df["construction_year"].replace('قبل از ۱۳۷۰',1370)
fa_to_en = str.maketrans("۰۱۲۳۴۵۶۷۸۹", "0123456789")
df["construction_year"] = (df["construction_year"].astype(str).str.translate(fa_to_en))
df["construction_year"] = pd.to_numeric(df["construction_year"],errors="coerce")
bins = np.arange(1370,1405,1)
plt.figure(figsize=(20,8))
plt.hist(df["construction_year"],bins=bins)
plt.xlabel("Construction Year")
plt.ylabel("Count")
plt.title("Distribution of Construction Year")
plt.xticks(bins,rotation=45)
plt.show()
```

# The table below shows the percentage growth in property construction compared to the previous :

```python
changes = df['construction_year'].value_counts().sort_index().pct_change()
changes[1370.0] = 0
print(changes * 100)
```
