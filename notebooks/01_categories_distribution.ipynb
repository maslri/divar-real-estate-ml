```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("divar.csv",low_memory=False)
```

```python
plt.figure(figsize=(18,9))
df["cat2_slug"].value_counts().plot(kind="bar")
plt.title("Distribution of Ads by Category Tier 2")
plt.xlabel("Category")
plt.ylabel("Number of Ads")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(18,9))
df["cat3_slug"].value_counts().plot(kind="bar")
plt.title("Distribution of Ads by Category Tier 3")
plt.xlabel("Category")
plt.ylabel("Number of Ads")
plt.xticks(rotation=80)
plt.show()
```

# Distributions of categories show that most of the ads are about apartment-sell and rent
